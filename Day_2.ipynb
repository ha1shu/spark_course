```python
from pyspark.sql.functions import *

df1_data = [
    ('Harshit','1'),
    ('Ravi','2'),
    ('suman','3'),
    ('vievk','4'),
    ('kumar','5')
]

schema1 = 'name string, id string'

df1 = spark.createDataFrame(df1_data, schema1)


df2_data = [
    (1,'Harst'),
    (2,'rohit'),
    (3,'sun'),
    (4,'vikas')
]

schema2 = 'id int, name string'

df2 = spark.createDataFrame(df2_data, schema2)

```

```python
# Convert datatype
df2 = df2.withColumn("id", col("id").cast("string"))

```

### Union

```python
df = df1.union(df2).display()
```

### unionByName

```python
# Union
df = df1.unionByName(df2)

display(df)
```
