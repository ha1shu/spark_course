```python
from pyspark.sql.functions import *
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.table('practice_catalog.new_source.big_mart_sales_1')
display(df)
```

```python
df.printSchema()
```

```python
df.filter(col('Item_Fat_Content') == 'Regular').display()
```

```python
df.filter((col('Item_Weight')<10) & (col('Item_Type')=='Snack Foods')).display()
```

```python
df.printSchema()
```

```python
df.filter((col('Outlet_Size').isNull()) & (col('Outlet_Location_Type').isin('Tier 1','Tier 2'))).display()
```

```python
df.withColumnRenamed('Item_Weight','Item_New_Weight').display()
```

```python
df = df.withColumn(
    'mul',
    (col('Item_Weight') * col('Item_MRP')).cast(IntegerType())
)
```

```python
df = df.withColumn('Item_Fat_Content',regexp_replace(col('Item_Fat_Content'),"Reg","Regular"))\
    .withColumn('Item_Fat_Content',regexp_replace(col('Item_Fat_Content'),"Low Fat","Low Fat"))
      
```

```python
df.display()
```

```python
df = df.drop('Item_Fat_Content')
      
```

```python
df.display()
```

```python
df.sort(["Item_Outlet_Sales","Item_Weight","Item_MRP"],ascending = [0,0,1]).display()
```
