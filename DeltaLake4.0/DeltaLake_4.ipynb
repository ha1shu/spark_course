```python
data = [(7,'Harsh',770),(5,'Ravi',1200),(6,'Raj',3300)]
df = spark.createDataFrame(data,["id","name","salary"])

df.write.format("delta")\
    .mode('append')\
    .save('/Volumes/deltalakefirstcatalog/default/deltalogvolume1/dmlSink1/')
```

```python
%sql
select * from delta .`/Volumes/deltalakefirstcatalog/default/deltalogvolume1/dmlSink1/`
```

```python
dlt_obj = DeltaTable.forPath(spark,'/Volumes/deltalakefirstcatalog/default/deltalogvolume1/dmlSink1/')

dlt_obj.alias('trg').merge(df.alias('src'),'trg.id = src.id')\
    .whenMatchedUpdateAll()\
    .whenNotMatchedInsertAll()\
    .execute()
```
