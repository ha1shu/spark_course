```python
%sql
create TABLE deltalakefirstcatalog.default.sqlFirstTable
(
    id INT NOT NULL,
    name string NOT NULL
)
```

```python
from delta.tables import DeltaTable , IdentityGenerator
```

```python
DeltaTable.createIfNotExists(spark)\
    .tableName('deltalakefirstcatalog.default.DeltaApiFirstTable')\
    .addColumn('id','int')\
    .addColumn('first_name','string')\
    .addColumn('last_name','string',comment='This is the last name')\
    .execute()
    
```

```python
%sql

drop table deltalakefirstcatalog.default.sqlFirstTable
```

```python
from pyspark.sql.types import LongType

DeltaTable.create(spark)\
  .tableName("deltalakefirstcatalog.default.deltaIDTable")\
  .addColumn("id_col", dataType=LongType(), generatedAlwaysAs=IdentityGenerator())\
  .addColumn("first_name", "string")\
  .addColumn("last_name", "string")\
  .execute()
```

```python
from pyspark.sql.types import *
```

```python
DeltaTable.createIfNotExists(spark)\
    .tableName('deltalakefirstcatalog.default.DeltaApiSecondTable')\
    .addColumn('id',dataType=LongType(),generatedAlwaysAs=IdentityGenerator())\
    .addColumn('first_name','string')\
    .addColumn('last_name','string')\
    .addColumn('Salary',dataType=IntegerType())\
    .addColumn('SalaryAfterTax',dataType=LongType(),generatedAlwaysAs='cast((Salary*0.9) as BIGINT)')\
    .execute()
    
```
