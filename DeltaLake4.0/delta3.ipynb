### **SchemaEvolution**

```python
data = [(1, 'John', 'USA'), (2, 'Jane', 'Canada'), (3, 'Bob', 'Mexico')]
df = spark.createDataFrame(data, ['id', 'name', 'country'])

display(df)
```

```python
df.write.format('delta')\
    .mode('overwrite')\
    .option('overwriteSchema', 'true')\
    .save('/Volumes/deltalakefirstcatalog/default/deltalogvolume1/DataSink/')
```

```python
%sql
SELECT *
FROM delta.`/Volumes/deltalakefirstcatalog/default/deltalogvolume1/DataSink`;
```
