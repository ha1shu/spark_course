```python
data = [(1,"Harshit","India"),(2,"Raj","USA"),(3,"Priya","India"),(4,"Ravi","USA"),(5,"Priya","India"),(6,"Ravi","USA")]
column = ["id","name","country"]
df = spark.createDataFrame(data,column)

display(df)
```

```python
df.write.format('delta')\
    .mode('append')\
    .save('/Volumes/deltalakefirstcatalog/default/deltavol1')
```
